## Problems with Perceptron

The main problem with perceptron which was reason that it didn't quite get famous in the Deep Learning is that it cannot converge on Non-linear data. It works very perfectly with linear data but for nonlinear it is just waste.

That is why it cannot solve a XOR logic problem. Here, I will demonstrate that---

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
or_dataset = pd.DataFrame()
and_dataset = pd.DataFrame()
xor_dataset = pd.DataFrame()
```

```python
or_dataset["input1"] = [1,1,0,0]
or_dataset["input2"] = [1,0,1,0]
or_dataset["output"] = [1,1,1,0]
```

```python
and_dataset["input1"] = [1,1,0,0]
and_dataset["input2"] = [1,0,1,0]
and_dataset["output"] = [1,0,0,0]
```

```python
xor_dataset["input1"] = [1,1,0,0]
xor_dataset["input2"] = [1,0,1,0]
xor_dataset["output"] = [0,1,1,0]
```

```python
or_dataset
```

```python
from sklearn.linear_model import Perceptron
clf1=Perceptron()
clf2=Perceptron()
clf3=Perceptron()
```

```python
clf1.fit(or_dataset.iloc[:,0:2].values,or_dataset.iloc[:,-1].values)
clf2.fit(and_dataset.iloc[:,0:2].values,and_dataset.iloc[:,-1].values)
clf3.fit(xor_dataset.iloc[:,0:2].values,xor_dataset.iloc[:,-1].values)
```

```python
clf2.coef_
```

```python
clf2.intercept_
```

```python
x=np.linspace(-1,1,5)
y=-x+1.5
```

```python
plt.plot(x,y)
plt.scatter(and_dataset["input1"], and_dataset["input2"], c=and_dataset["output"], s=200)
```

```python
x1=np.linspace(-1,1,5)
y1=-x1+0.5
```

```python
plt.plot(x1,y1)
plt.scatter(or_dataset["input1"], or_dataset["input2"], c=or_dataset["output"], s=200)
```
