## Gradient Descent Example

```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
X,y = make_regression(n_samples=10, n_features=1, n_targets=1, n_informative=1, noise=80, random_state=13)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X,y)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
model.fit(X,y)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
plt.scatter(X,y)
plt.plot(X,model.predict(X),color='green')
```

```python
# Let's apply Gradient Descent assuming slope is 48.53048884
# Let's intial value of b=0
y_pred = ((48.53048884 * X) + 0).reshape(10)
```

```python
plt.scatter(X,y)
plt.plot(X,model.predict(X),color='red',label='OLS') 
plt.plot(X,y_pred,color="#279406",label='b = 0')
plt.grid()
plt.legend()
plt.show() 
```

```python
m = 48.53048884
b = 0

# Iteration 1
loss_slope = -2 * np.sum(y - m*X.ravel() - 0)
loss_slope
```

```python
# let's take learning rate = 0.1
lr = 0.1
step_size = loss_slope * lr
step_size
```

```python
b = b - step_size
b
```

```python
y_pred1 = ((48.53048884 * X) + b).reshape(10)

plt.scatter(X,y)
plt.plot(X,model.predict(X),color='red',label='OLS') 
plt.plot(X,y_pred,color="#279406",label='b = 0')
plt.plot(X,y_pred1,color="#1AA2D8",label='b = {}'.format(b))
plt.grid()
plt.legend()
plt.show() 
```

```python
 # Iteration
loss_slope = -2 * np.sum(y - m*X.ravel() - b)
loss_slope
```

```python
step_size = lr * loss_slope
step_size
```

```python
b = b - step_size
b
```

```python
y_pred2 = ((48.53048884 * X) + b).reshape(10)

plt.scatter(X,y)
plt.plot(X,model.predict(X),color='red',label='OLS') 
plt.plot(X,y_pred,color="#279406",label='b = 0')
plt.plot(X,y_pred1,color="#1AA2D8",label='b = {}'.format(b))
plt.plot(X,y_pred2,color="#F6F616",label='b = {}'.format(b))
plt.grid()
plt.legend()
plt.show() 
```

```python
b = -100
m = 48.53
lr = 0.1
epochs = 10

for i in range(epochs):
    loss_slope = -2 * np.sum(y - m*X.ravel() - b)
    b = b - (lr*loss_slope)

    y_pred = m*X + b
    plt.plot(X, y_pred)

plt.scatter(X,y)
plt.grid()
```
