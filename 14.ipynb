## Memoization in Backpropagation

```python
def fib(n):
    if n==0 or n==1:
        return n
    else:
        return fib(n-1) + fib(n-2)
```

```python
fib(36)
```

```python
# By Memoization
d = {0:0, 1:1}
def fib(n,d):
    if n in d:
        return d[n]
    else: 
        d[n] = fib(n-1,d) + fib(n-2,d)
        return d[n]
```

```python
fib(36,d)
```
